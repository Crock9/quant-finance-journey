## 1. Create a price series

Imagine these are the closing prices of one stock over five trading days.

````python
prices = [100, 102, 101, 105, 107]
print(prices)
````

## 2. Calculate one return by hand

The simple-return formula is:

$$\text{return} = \frac{\text{new price}}{\text{old price}} - 1$$

Multiplying by 100 expresses the result as a percentage.

````python
first_return = (prices[1] / prices[0]) - 1
print(f"First daily return: {first_return * 100:.2f}%")
````

## 3. Calculate every daily return

The loop begins at position 1 because every return needs both today's price and the previous day's price.

````python
returns = []

for i in range(1, len(prices)):
    daily_return = (prices[i] / prices[i - 1]) - 1
    returns.append(daily_return)
    print(f"Return {i}: {daily_return * 100:.2f}%")
````

## 4. Find the average daily return

````python
average_return = sum(returns) / len(returns)
print(f"Average daily return: {average_return * 100:.2f}%")
````
